<b> 2. Analyse et traitement des valeurs manquantes

```python
# --- Nettoyage Category Code ---
df_clean = df.copy()

# Remplacer les strings vides/blancs par NaN pour category_code
df_clean['category_code'] = df_clean['category_code'].replace(r'^\s*$', pd.NA, regex=True)

# Remplacer NaN par le code inconnu
df_clean['category_code'] = df_clean['category_code'].fillna("unknown_category_lvl1.unknown_category_lvl2.unknown_category_lvl3")
```

```python
# Supprimer les lignes où user_session est manquant (NaN ou vide ou petites valeurs)
df_clean = df_clean[
    df_clean['user_session'].notna() &                # Pas NaN
    (df_clean['user_session'].astype(str).str.len() > 5)  # Au moins une session valide (ex: UUID long)
]

print("Shape après suppression des mauvaises user_session :", df_clean.shape)
```

```python
# --- Nettoyage Brand ---

# Remplacer les strings vides/blancs par NaN pour brand
df_clean['brand'] = df_clean['brand'].replace(r'^\s*$', pd.NA, regex=True)

# Remplacer NaN par "unknown_brand"
df_clean['brand'] = df_clean['brand'].fillna("unknown_brand")
df_clean.head()
```

```python
# --- Valeurs manquantes ---
missing_values = df_clean.isnull().sum()
missing_values = missing_values[missing_values > 0]
missing_values
```

```python
# --- Filtrage des prix aberrants ---
lower_bound = 0
upper_bound = df_clean['price'].quantile(0.99)
print(f"Nombre de lignes avant nettoyage : {df_clean.shape[0]}")

# Visualiser les prix avec Plotly Scatter Plot
fig_scatter = px.scatter(
    df_clean.reset_index(),
    x=df_clean.index,
    y='price',
    title="Visualisation des prix avec seuils aberrants",
    labels={'index': "Index de l'enregistrement", 'price': "Prix du produit"}
)
fig_scatter.add_hline(y=upper_bound, line_dash="dash", line_color="red", annotation_text="Seuil 99%")
fig_scatter.add_hline(y=lower_bound, line_dash="dash", line_color="green", annotation_text="0€")
fig_scatter.show()

# Nettoyage effectif
# df_clean = df_clean[(df_clean['price'] >= lower_bound) & (df_clean['price'] <= upper_bound)]

print(f"Nombre de lignes après nettoyage : {df_clean.shape[0]}")
```

<b>7. Feature Engineering : Construction des variables comportementales

```python
# --- On utilise le dataset nettoyé ---
df = df_clean.copy()

# --- Séparation des événements par type ---
views = df[df['event_type'] == 'view']
purchases = df[df['event_type'] == 'purchase']

# --- Calcul des statistiques de base ---
features = df.groupby('user_id').agg(
    total_events=('event_type', 'count'),
    n_sessions=('user_session', pd.Series.nunique),
    n_views=('event_type', lambda x: (x == 'view').sum()),
    n_carts=('event_type', lambda x: (x == 'cart').sum()),
    n_purchases=('event_type', lambda x: (x == 'purchase').sum())
).reset_index()
features.head()
```

```python
# --- Ratio achats / vues ---
features['purchase_view_ratio'] = features['n_purchases'] / features['n_views']
features['purchase_view_ratio'] = features['purchase_view_ratio'].fillna(0)  # Remplacer les NaN éventuels
features.head()
```

```python
# --- Prix des produits vus et achetés ---
avg_price_viewed = views.groupby('user_id')['price'].mean().rename('avg_price_viewed')
avg_price_purchased = purchases.groupby('user_id')['price'].mean().rename('avg_price_purchased')
max_price_purchased = purchases.groupby('user_id')['price'].max().rename('max_price_purchased')
min_price_purchased = purchases.groupby('user_id')['price'].min().rename('min_price_purchased')

# Fusion avec le dataset features
features = features.merge(avg_price_viewed, on='user_id', how='left')
features = features.merge(avg_price_purchased, on='user_id', how='left')
features = features.merge(max_price_purchased, on='user_id', how='left')
features = features.merge(min_price_purchased, on='user_id', how='left')

# --- Diversité des catégories ---
n_unique_categories_viewed = views.groupby('user_id')['category_lvl1'].nunique().rename('n_unique_categories_viewed')
n_unique_categories_purchased = purchases.groupby('user_id')['category_lvl1'].nunique().rename('n_unique_categories_purchased')

features = features.merge(n_unique_categories_viewed, on='user_id', how='left')
features = features.merge(n_unique_categories_purchased, on='user_id', how='left')
features.head()
```

```python
# --- Catégories préférées ---
favorite_category_viewed = views.groupby('user_id')['category_lvl1'].agg(lambda x: x.value_counts().idxmax()).rename('favorite_category_viewed')
favorite_category_purchased = purchases.groupby('user_id')['category_lvl1'].agg(lambda x: x.value_counts().idxmax()).rename('favorite_category_purchased')

features = features.merge(favorite_category_viewed, on='user_id', how='left')
features = features.merge(favorite_category_purchased, on='user_id', how='left')

# --- Marque préférée ---
favorite_brand_viewed = views.groupby('user_id')['brand'].agg(lambda x: x.value_counts().idxmax()).rename('favorite_brand_viewed')

features = features.merge(favorite_brand_viewed, on='user_id', how='left')

# --- Aperçu final ---
print("Aperçu du dataset utilisateur enrichi :")
features.head()
```

<b> Réduction de dimension avec PCA (Analyse en Composantes Principales)

```python
# --- Imports ---
from sklearn.decomposition import PCA
import plotly.express as px
import pandas as pd

# --- Sélectionner les colonnes numériques finales à passer dans le PCA ---
pca_features = [
    'total_events', 'n_sessions', 'n_views', 'n_purchases',
    'purchase_view_ratio', 'avg_price_viewed', 'avg_price_purchased',
    'max_price_purchased', 'min_price_purchased',
    'n_unique_categories_viewed', 'n_unique_categories_purchased',
    'favorite_category_viewed', 'favorite_category_purchased', 'favorite_brand_viewed'
]

# --- Appliquer le PCA ---
pca = PCA(n_components=2)

# Fit et transform
principal_components = pca.fit_transform(features_scaled[pca_features])

# --- Créer un DataFrame avec les nouvelles composantes ---
pca_df = pd.DataFrame(principal_components, columns=['PC1', 'PC2'])

# --- Afficher la variance expliquée ---
explained_variance = pca.explained_variance_ratio_
print("Variance expliquée par chaque composante principale :")
for i, var in enumerate(explained_variance):
    print(f"PC{i+1} : {var:.2%}")

print(f"Variance cumulée par PC1 + PC2 : {(explained_variance[0] + explained_variance[1]):.2%}")
print(f"Nombre d'observations après PCA : {pca_df.shape[0]}")

# --- Visualisation 2D (PC1 vs PC2) ---
fig_2d = px.scatter(
    pca_df,
    x='PC1',
    y='PC2',
    title='Projection 2D PCA : PC1 vs PC2',
    labels={'PC1': 'Composante Principale 1', 'PC2': 'Composante Principale 2'},
    opacity=0.7
)
fig_2d.show()
```

<b> Build du model: KMeans

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import plotly.express as px

# --- Déterminons le nombre optimal de clusters automatiquement ---

inertia = []
silhouette_scores = []
k_values = range(2, 5)  # Tester de 2 à 4 clusters

for k in k_values:
    print(f" Traitement pour k={k} clusters...")

    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10, verbose=0)
    kmeans.fit(pca_df[['PC1', 'PC2']])

    current_inertia = kmeans.inertia_
    current_silhouette = silhouette_score(pca_df[['PC1', 'PC2']], kmeans.labels_)

    inertia.append(current_inertia)
    silhouette_scores.append(current_silhouette)

    print(f" k={k} terminé : Inertia={current_inertia:.2f}, Silhouette={current_silhouette:.4f}")
    print("-" * 50)

print("Tous les k ont été traités.")

# --- Choix automatique du k avec meilleur silhouette score ---
best_k = k_values[silhouette_scores.index(max(silhouette_scores))]
print(f"\n Le meilleur nombre de clusters est k={best_k} avec un silhouette score de {max(silhouette_scores):.4f}")
```

```python
# --- Visualisation méthode du coude ---
plt.figure(figsize=(10,5))
plt.plot(k_values, inertia, marker='o')
plt.title('Méthode du Coude : Inertie en fonction du nombre de clusters')
plt.xlabel('Nombre de clusters')
plt.ylabel('Inertie')
plt.grid(True)
plt.show()

# --- Visualisation silhouette scores ---
plt.figure(figsize=(10,5))
plt.plot(k_values, silhouette_scores, marker='x', color='orange')
plt.title('Silhouette Score en fonction du nombre de clusters')
plt.xlabel('Nombre de clusters')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans
import plotly.express as px

optimal_k = best_k 

# --- Appliquons KMeans final ---
kmeans_final = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10,
    verbose=1
)
pca_df['cluster_label'] = kmeans_final.fit_predict(pca_df[['PC1', 'PC2']])
```

```python
# ---Visaualisation les clusters sur PC1 vs PC2 ---
fig_clusters = px.scatter(
    pca_df,
    x='PC1',
    y='PC2',
    color='cluster_label',
    title=f'Clusters des utilisateurs projetés sur PCA (k={optimal_k})',
    labels={'PC1': 'Composante 1', 'PC2': 'Composante 2', 'cluster_label': 'Cluster'},
    opacity=0.7
)
fig_clusters.show()
```

```python
# --- Ajout des labels de cluster dans features ---
features_with_clusters = features_scaled.copy()
features_with_clusters['cluster_label'] = pca_df['cluster_label']

# --- Analyse des clusters ---
cluster_analysis = features_with_clusters.groupby('cluster_label').mean()

# --- Affichage proprement ---
print("Analyse des clusters :")
cluster_analysis
```

<b> Calcule des métriques

```python
# Évalue la qualité d'un clustering en affichant 3 scores clés :
# - Silhouette Score (plus haut est meilleur)
# - Calinski-Harabasz Score (plus haut est meilleur)
# - Davies-Bouldin Score (plus bas est meilleur)
```

```python
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
import pandas as pd

# --- Variables utilisées pour le clustering ---
X = pca_df[['PC1', 'PC2']]  # Espace réduit à 2D par PCA

# --- Préparer les étiquettes de chaque modèle ---
labels_dict = {
    'KMeans': pca_df['cluster_label'],
}

# --- Initialiser une liste pour stocker les résultats ---
results = []

# --- Calculer les métriques pour chaque modèle ---
for model_name, labels in labels_dict.items():
    X_filtered = X
    labels_filtered = labels
    silhouette = silhouette_score(X_filtered, labels_filtered)
    calinski = calinski_harabasz_score(X_filtered, labels_filtered)
    davies = davies_bouldin_score(X_filtered, labels_filtered)

    results.append({
        'Modèle': model_name,
        'Silhouette Score': round(silhouette, 4),
        'Calinski-Harabasz Score': round(calinski, 2),
        'Davies-Bouldin Score': round(davies, 4)
    })

# --- Résultats sous forme de tableau ---
results_df = pd.DataFrame(results)

results_df
```

```python
def predict_user_batch(raw_df):
    """
    Prend en entrée un DataFrame brut de nouveaux utilisateurs,
    retourne leurs features + cluster prédits.
    """

    from sklearn.preprocessing import LabelEncoder

    # --- Feature Engineering ---
    new_features = raw_df.groupby('user_id').agg(
        total_events=('event_type', 'count'),
        n_sessions=('user_session', pd.Series.nunique),
        n_views=('event_type', lambda x: (x == 'view').sum()),
        n_purchases=('event_type', lambda x: (x == 'purchase').sum())
    ).reset_index()

    new_features['purchase_view_ratio'] = new_features['n_purchases'] / new_features['n_views']
    new_features['purchase_view_ratio'] = new_features['purchase_view_ratio'].fillna(0)

    views = raw_df[raw_df['event_type'] == 'view']
    purchases = raw_df[raw_df['event_type'] == 'purchase']

    avg_price_viewed = views.groupby('user_id')['price'].mean().rename('avg_price_viewed')
    avg_price_purchased = purchases.groupby('user_id')['price'].mean().rename('avg_price_purchased')
    max_price_purchased = purchases.groupby('user_id')['price'].max().rename('max_price_purchased')
    min_price_purchased = purchases.groupby('user_id')['price'].min().rename('min_price_purchased')

    new_features = new_features.merge(avg_price_viewed, on='user_id', how='left')
    new_features = new_features.merge(avg_price_purchased, on='user_id', how='left').fillna(0)
    new_features = new_features.merge(max_price_purchased, on='user_id', how='left').fillna(0)
    new_features = new_features.merge(min_price_purchased, on='user_id', how='left').fillna(0)

    new_features['n_unique_categories_viewed'] = views.groupby('user_id')['category_code'].nunique()
    new_features['n_unique_categories_purchased'] = purchases.groupby('user_id')['category_code'].nunique().fillna(0)

    for col in ['favorite_category_viewed', 'favorite_category_purchased', 'favorite_brand_viewed']:
        new_features[col] = 'unknown'

    for col in ['favorite_category_viewed', 'favorite_category_purchased', 'favorite_brand_viewed']:
        le = LabelEncoder()
        new_features[col] = le.fit_transform(new_features[col])

    # --- Remplacer NaN par 0 avant normalisation ---
    new_features = new_features.fillna(0)

    # --- Remplacer les inf et -inf par NaN ---
    new_features = new_features.replace([np.inf, -np.inf], np.nan)
    
    # --- Remplacer tous les NaN par 0 ---
    new_features = new_features.fillna(0)

    # --- Normalisation et PCA ---
    new_features_scaled = scaler.transform(new_features.drop(columns=['user_id']))
    new_features_pca = pca.transform(new_features_scaled)[:, :2]

    # --- Prédiction ---
    predicted_clusters = kmeans_final.predict(new_features_pca)

    # --- Ajouter la colonne des clusters aux features d'origine ---
    new_features['final_cluster_label'] = predicted_clusters

    return new_features  # <--- On retourne toutes les features + label
```

```python
# Label que tu conserves =>	Ce qu'il évoque
# VIP Clients	Top clients => priorité business
# Grands Acheteurs =>	Clients qui dépensent beaucoup
# Acheteurs Curieux =>	Clients qui explorent beaucoup de produits avant d'acheter
# Acheteurs Réguliers	=> Clients qui achètent fréquemment
# Explorateurs Intenses =>	Visiteurs qui visitent plein de produits sans acheter
# Explorateurs	=> Visiteurs qui explorent un peu sans achat
# Visiteurs Inactifs =>	Presque aucune activité
# Clients Classiques =>	Moyenne des comportements
```

```python
def generate_dynamic_cluster_labels(features_with_clusters):
    """
    Génère un tableau dynamique d'interprétation des clusters
    avec des labels clairs et expressifs.

    Args:
        features_with_clusters: DataFrame avec au moins :
            - 'final_cluster_label'
            - 'purchase_view_ratio'
            - 'avg_price_purchased'
            - 'n_unique_categories_viewed'

    Returns:
        DataFrame contenant pour chaque cluster :
            - purchase_view_ratio (moyenne)
            - avg_price_purchased (moyenne)
            - n_unique_categories_viewed (moyenne)
            - label attribué automatiquement
    """

    # --- Calcul des moyennes par cluster ---
    cluster_summary = features_with_clusters.groupby('final_cluster_label').agg({
        'purchase_view_ratio': 'mean',
        'avg_price_purchased': 'mean',
        'n_unique_categories_viewed': 'mean'
    }).reset_index()

    # --- Définir la fonction de labellisation expressive ---
    def label_cluster(row):
        ratio = row['purchase_view_ratio']
        avg_price = row['avg_price_purchased']
        n_cat = row['n_unique_categories_viewed']

        if ratio > 2 and avg_price > 2:
            return "VIP Clients"
        elif ratio > 1.5 and avg_price > 1.5:
            return "Gros Acheteurs"
        elif ratio > 1.5 and n_cat > 2:
            return "Acheteurs Curieux"
        elif ratio > 1:
            return "Acheteurs Réguliers"
        elif ratio < 0.5 and n_cat > 3:
            return "Explorateurs Intenses"
        elif ratio < 0.5 and n_cat > 1:
            return "Explorateurs"
        elif abs(ratio) < 0.3 and abs(avg_price) < 0.3 and abs(n_cat) < 0.3:
            return "Visiteurs Inactifs"
        else:
            return "Clients Classiques"

    cluster_summary['label'] = cluster_summary.apply(label_cluster, axis=1)

    return cluster_summary
```

```python
df_test = pd.read_csv('../dataset_2.csv',nrows=20)
```

```python
# Feature Engineering + Clustering
features_with_clusters = predict_user_batch(df_test)

# Génération dynamique des labels par cluster
dynamic_profiles = generate_dynamic_cluster_labels(features_with_clusters)

# Joindre le label interprété à chaque utilisateur
# (Merge sur 'final_cluster_label')
final_result = features_with_clusters.merge(dynamic_profiles[['final_cluster_label', 'label']], on='final_cluster_label', how='left')

# Affichage final
final_result[['user_id', 'final_cluster_label', 'label']]
```
